```python
import cv2
import os
import numpy as np
import shutil
from ultralytics import SAM
from google.colab import files

# 1. Create a clean folder for pure binary masks
out_dir = '/content/pure_masks'
if os.path.exists(out_dir):
    shutil.rmtree(out_dir)
os.makedirs(out_dir, exist_ok=True)

# 2. Run SAM on the evaluation split (without saving the visualization overlays)
model = SAM('sam_b.pt')
results = model('/content/dataset/valid/egg')

# 3. Extract the raw mathematical boundaries and draw perfect black/white masks
for result in results:
    filename = os.path.basename(result.path)
    orig_shape = result.orig_shape # Get original (Height, Width)

    # Create a pure black canvas
    binary_mask = np.zeros(orig_shape, dtype=np.uint8)

    # If the AI found an object, draw it in solid white
    if result.masks is not None:
        for xy in result.masks.xy: # .xy contains the polygon coordinates
            pts = np.int32([xy])
            cv2.fillPoly(binary_mask, pts, 255)

    cv2.imwrite(os.path.join(out_dir, filename), binary_mask)

# 4. Zip and download
!zip -r /content/pure_masks.zip /content/pure_masks
files.download('/content/pure_masks.zip')
```
